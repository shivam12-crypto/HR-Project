```python
# cleaning data set
# removing unnamed column
df.drop(columns="Unnamed: 0",inplace=True)
```

```python
df.head()
```

```python
# converting hire date to date time and seperating it
df["Hire_Date"] = pd.to_datetime(df["Hire_Date"])
df["Hire_Day"] = df["Hire_Date"].dt.day
df["Hire_Month"] = df["Hire_Date"].dt.month
df["Hire_Year"] = df["Hire_Date"].dt.year
```

```python
df.head()
```

```python
# removing the hire date column
df.drop(columns="Hire_Date",inplace=True)
```

```python
df.head()
```

```python
# seperating categorical and numerical data
df_cat = df.select_dtypes(include="object")
df_num = df.select_dtypes(include=["int","float"])
```

```python
# visualization
df_cat.head()
```

```python
# department and job title of employees
plt.figure(figsize=(15,6))

plt.subplot(1,2,1)
dt = df_cat["Department"].value_counts()
plt.title("Department of Employees")
sns.barplot(data=dt,palette="ocean")
sns.lineplot(data=dt,marker="^")
plt.xlabel("Department")
plt.xticks(rotation=90)
plt.ylabel("Number of Employees")

plt.subplot(1,2,2)
dt = df_cat["Job_Title"].value_counts()
plt.title("Job Title of Employees")
sns.barplot(data=dt,palette="Set1")
sns.lineplot(data=dt,marker="^")
plt.xlabel("Job Title")
plt.xticks(rotation=90)
plt.ylabel("Number of Employees")
plt.show()

plt.tight_layout()
```

```python
# department and job title of active employees
plt.figure(figsize=(15,8))

plt.subplot(1,2,1)
dt = df_cat[df_cat["Status"] == "Active"]["Department"].value_counts()
plt.title("Department of Active Employees")
sns.barplot(data=dt,palette="hls")
plt.xlabel("Department")
plt.xticks(rotation=90)
plt.ylabel("Number of employees")

plt.subplot(1,2,2)
dt = df_cat[df_cat["Status"] == "Active"]["Job_Title"].value_counts()
plt.title("Job Title of Active Employees")
sns.barplot(data=dt,palette="Reds")
plt.xlabel("Job Title")
plt.xticks(rotation=90)
plt.ylabel("Number of employees")

plt.tight_layout()
```

```python
# departemt and job title of retired employees
plt.figure(figsize=(15,10))

plt.subplot(1,2,1)
dt = df_cat[df_cat["Status"] == "Retired"]["Department"].value_counts()
plt.title("Department of Employees")
percentage = (df_cat[df_cat["Status"] == "Retired"]["Department"].value_counts()/df_cat[df_cat["Status"] == "Retired"]["Department"].value_counts().sum())*100
plt.pie(data=dt,x=percentage,autopct="%1.2f")
plt.legend(dt.index,)

plt.subplot(1,2,2)
dt_1 = df_cat[df_cat["Status"] == "Retired"]["Job_Title"].value_counts()
plt.title("Job Title Employees")
sns.barplot(data=dt_1,palette="Set2")
sns.lineplot(data=dt_1,marker="o",color="orange")
plt.xlabel("Job Title")
plt.xticks(rotation=90)
plt.ylabel("Number of Employees")

plt.tight_layout()
```

```python
# status and work mode of employees
for item in df_cat.columns:
    if item == "Status" or item == "Work_Mode":
        dt = df_cat[item].value_counts()
        percentage = (df_cat[item].value_counts()/df_cat[item].value_counts().sum())*100
        plt.figure(figsize=(8,8))
        plt.title(f"Employee's {item}")
        plt.pie(data=df_cat,x=percentage,autopct="%1.2f")
        plt.legend(dt.index)
        plt.show()
```

```python
# most prominent locations from where employees are on board
location = df_cat[df_cat["Status"] == "Active"]["Location"].value_counts()[:10]
percentage = (df_cat[df_cat["Status"] == "Active"]["Location"].value_counts()[:10]/df_cat["Location"].value_counts()[:10].sum())*100
plt.figure(figsize=(16,16))
plt.title("Prominent Hiring Locations")
plt.pie(data=df_cat,x=percentage,autopct="%1.2f")
plt.legend(location.index)
plt.show()
```

```python
# daily hirings
plt.figure(figsize=(20,4))

plt.subplot(1,3,1)
dt = df_num["Hire_Day"].value_counts()
plt.title("Hirings on Daily Basis")
sns.lineplot(data=dt,color="green")
plt.xlabel("Day")
plt.ylabel("Number of Employees")

plt.subplot(1,3,2)
dt = df_num["Hire_Month"].value_counts()
plt.title("Hirings on Monthly Basis")
sns.lineplot(data=dt,color="red")
plt.xlabel("Month")
plt.ylabel("Number of Employees")

plt.subplot(1,3,3)
dt = df_num["Hire_Year"].value_counts()
plt.title("Hirings on Yearly Basis")
sns.lineplot(data=dt,color="purple")
plt.xlabel("Year")
plt.ylabel("Number of Employees")

plt.suptitle("Employee Hirings")
plt.tight_layout()
```

```python
# employee performance rating
rating = df_num["Performance_Rating"].value_counts()
percentage = (df_num["Performance_Rating"].value_counts()/df_num["Performance_Rating"].value_counts().sum())*100
plt.figure(figsize=(6,6))
plt.title("Employee Performance Rating")
plt.pie(data=df_num,x=percentage,autopct="%1.2f")
plt.legend(rating.index)
plt.show()
```

```python
# employee experiance
experiance = df_num["Experience_Years"].value_counts()
plt.title("Employee's Experiance")
sns.barplot(data=experiance,palette="viridis")
sns.lineplot(data=experiance,marker="o")
plt.xlabel("Years of Experiance")
plt.ylabel("Number of Employees")
plt.show()
```

```python
# employee salary
plt.title("Employee Salary")
sns.histplot(data=df_num,x=df_num["Salary_INR"],color="purple",kde=True)
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```
